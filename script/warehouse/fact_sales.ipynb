```python
import pandas as pd

def run():
    orders = pd.read_csv('/opt/airflow/data/raw/olist_orders_dataset.csv')
    items = pd.read_csv('/opt/airflow/data/raw/olist_order_items_dataset.csv')

    fact = items.merge(orders, on='order_id')

    fact.to_csv('/opt/airflow/data/dwh/fact_sales.csv', index=False)

    print("fact_sales created")

if __name__ == "__main__":
    run()
```
