```python
import pandas as pd

def run():
    fact = pd.read_csv('/opt/airflow/data/dwh/fact_sales.csv')
    product = pd.read_csv('/opt/airflow/data/dwh/dim_product.csv')

    enriched = fact.merge(product, on='product_id', how='left')

    enriched.to_csv('/opt/airflow/data/dwh/fact_sales_enriched.csv', index=False)

    print("fact_sales_enriched created")

if __name__ == "__main__":
    run()
```
