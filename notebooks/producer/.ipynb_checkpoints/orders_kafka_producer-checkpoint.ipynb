# Orders_Item Kafka Producer

## 1. Start Spark

```python
from pyspark.sql import SparkSession

spark = (
    SparkSession.builder
    .appName('olist-kafka-producer')
    .master('local[*]')
    .config(
        'spark.jars.packages',
        'org.apache.spark:spark-sql-kafka-0-10_2.12:3.5.1,org.postgresql:postgresql:42.7.3'
    )
    .getOrCreate()
)

spark
```

```python
import os

DATA_PATH = os.getenv('DATA_PATH')
RAW_PATH = os.getenv('RAW_PATH')

print('DATA_PATH =', DATA_PATH)
print('RAW_PATH =', RAW_PATH)
```

```python
order_items_df = spark.read.csv(
    f"{RAW_PATH}/items/olist_order_items_dataset.csv",
    header=True,
    inferSchema=True
)

orders_df.printSchema()
orders_df.show(5, truncate=False)
```

# 3. Clean and cast

```python
from pyspark.sql import functions as F

order_items_clean_df = (
    order_items_df
    .select(
        F.col('order_id').cast('string').alias('order_id'),
        F.col('order_item_id').cast('int').alias('order_item_id'),
        F.col('product_id').cast('string').alias('product_id'),
        F.col('seller_id').cast('string').alias('seller_id'),
        F.to_timestamp('shipping_limit_date').alias('shipping_limit_date'),
        F.col('price').cast('double').alias('price'),
        F.col('freight_value').cast('double').alias('freight_value')
    )
)

order_items_clean_df.printSchema()
```

# 4. Validate and deduplicate

```python
order_items_valid_df = (
    order_items_clean_df
    .filter(F.col('order_id').isNotNull())
    .filter(F.col('order_item_id').isNotNull())
    .filter(F.col('product_id').isNotNull())
    .filter(F.col('seller_id').isNotNull())
    .filter(F.col('price').isNotNull())
    .filter(F.col('freight_value').isNotNull())
    .filter(F.col('price') >= 0)
    .filter(F.col('freight_value') >= 0)
    .dropDuplicates(['order_id', 'order_item_id'])
)

print('valid deduplicated order_items count:', order_items_valid_df.count())
```

# 5. Order by shipping date

```python
order_items_kafka_df = (
    order_items_valid_df
    .orderBy(F.col('shipping_limit_date').asc_nulls_last(), F.col('order_id').asc())
)

order_items_kafka_df.show(5, truncate=False)
```

# 6. Convert Spark rows to Python dicts

```python
order_items_for_kafka = [
    {
        'order_id': row['order_id'],
        'order_item_id': int(row['order_item_id']) if row['order_item_id'] is not None else None,
        'product_id': row['product_id'],
        'seller_id': row['seller_id'],
        'shipping_limit_date': row['shipping_limit_date'].isoformat() if row['shipping_limit_date'] else None,
        'price': float(row['price']) if row['price'] is not None else None,
        'freight_value': float(row['freight_value']) if row['freight_value'] is not None else None,
    }
    for row in order_items_kafka_df.collect()
]

print('records prepared for kafka:', len(order_items_for_kafka))
order_items_for_kafka[0]
```

# 7. Create Kafka producer

```python
import json
from kafka import KafkaProducer

producer = KafkaProducer(
    bootstrap_servers='kafka:9092',
    value_serializer=lambda v: json.dumps(v).encode('utf-8'),
    key_serializer=lambda v: v.encode('utf-8') if v else None,
)

print('Producer ready')
```

# 8. Send a small sample first

```python
sample_n = 10

for record in order_items_for_kafka[:sample_n]:
    message_key = f"{record['order_id']}_{record['order_item_id']}"
    producer.send('olist.order_items', key=message_key, value=record)

producer.flush()
print(f'Sent {sample_n} sample messages')
```

# 9. Send all messages

```python
for i, record in enumerate(order_items_for_kafka, start=1):
    message_key = f"{record['order_id']}_{record['order_item_id']}"
    producer.send('olist.order_items', key=message_key, value=record)

    if i % 5000 == 0:
        producer.flush()
        print(f'Sent {i} messages')

producer.flush()
print(f'All messages sent: {len(order_items_for_kafka)}')
```

# 10. Close producer

```python
producer.close()
print('Producer closed')
```
